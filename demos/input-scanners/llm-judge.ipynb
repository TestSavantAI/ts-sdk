```python
from testsavant.guard import InputGuard, OutputGuard
from testsavant.guard.input_scanners import LLMJudge
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
# medical sentiment analysis

# samples of highly medical text containing medical terminologies to express sentiment and their labels
examples = {
    "The patient's prognosis is poor due to advanced-stage cancer.": "negative",
    "The treatment plan has been effective, and the patient is showing signs of improvement.": "positive",
    "The diagnosis is uncertain, and further tests are needed to determine the condition.": "neutral",
    "The surgery was successful, and the patient is recovering well.": "positive",
    "The symptoms are worsening, and the patient is experiencing increased pain.": "negative"
}
guard = InputGuard()
invisible_text_scanner = LLMJudge(
    tag="guard-3",
    num_retry=3
                            )
guard.add_scanner(invisible_text_scanner)
corrects = []
for text, label in examples.items():
    result = guard.scan(text)
    # print(f"Input Text: {text}")
    print(f"Expected Label: {label}, got {result.is_valid}")
    # print(f"Confidence: {result.is_valid}")
    corrects.append((result.is_valid and label in ["neutral", "positive"]) or (not result.is_valid and label == "negative"))

acc = sum(corrects) / len(corrects)
print(f"Accuracy: {acc * 100:.2f}%")
```
