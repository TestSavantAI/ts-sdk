```python
from testsavant.guard import InputGuard, OutputGuard
from testsavant.guard.input_scanners import Regex
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
text1 = "Once upon a time, there were people"
text2 = "Once upon a time, there was one person"
text3 = "Here is my token: Bearer abc-def_123"
```

```python
guard = InputGuard()
regex_scanner = Regex(
    tag="default",
    patterns=[r"Bearer [A-Za-z0-9-._~+/]+"],
    case_sensitive=False,
    redact=True,
    is_blocked=False,
    )
guard.add_scanner(regex_scanner)
```

```python
for text in [text1, text2, text3]:
    result = guard.scan(text)
    if result.is_valid:
        print("LLM api called", result.sanitized_prompt)
    else:
        print("User request blocked")
```
