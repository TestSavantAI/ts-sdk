```python
from testsavant.guard import InputGuard, OutputGuard
from testsavant.guard.input_scanners import TokenLimit
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
text1 = "Once upon a time, there were people"
text2 = "Once upon a time, there was one person"
text3 = "Once upon a time, there were individuals"
```

```python
guard = InputGuard()
token_limit_scanner = TokenLimit(
    tag="default",
    limit=8,
    )
guard.add_scanner(token_limit_scanner)
```

```python
for text in [text1, text2, text3]:
    result = guard.scan(text)
    if result.is_valid:
        print("LLM api called")
    else:
        print("Input blocked")
```
