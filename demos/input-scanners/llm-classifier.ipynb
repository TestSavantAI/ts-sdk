```python
from testsavant.guard import InputGuard, OutputGuard
from testsavant.guard.input_scanners import LLM
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
# medical sentiment analysis

# samples of highly medical text containing medical terminologies to express sentiment and their labels
examples = {
    "The patient's prognosis is poor due to advanced-stage cancer.": "negative",
    "The treatment plan has been effective, and the patient is showing signs of improvement.": "positive",
    "The diagnosis is uncertain, and further tests are needed to determine the condition.": "neutral",
    "The surgery was successful, and the patient is recovering well.": "positive",
    "The symptoms are worsening, and the patient is experiencing increased pain.": "negative"
}
guard = InputGuard()
invisible_text_scanner = LLM(
    tag="default",
    instruction="""You are a medical expert. Given the medical text, classify it as valid if it has a positive or neutral sentiment according to medical success, and invalid if it has a negative medical sentiment.""",
    num_retry=3
                            )
guard.add_scanner(invisible_text_scanner)
corrects = []
for text, label in examples.items():
    result = guard.scan(text)
    # print(f"Input Text: {text}")
    print(f"Expected Label: {label}, got {result.is_valid}")
    # print(f"Confidence: {result.is_valid}")
    corrects.append((result.is_valid and label in ["neutral", "positive"]) or (not result.is_valid and label == "negative"))

acc = sum(corrects) / len(corrects)
print(f"Accuracy: {acc * 100:.2f}%")
```

```python
# finance sentiment analysis
finance_examples = {
    "The company's quarterly earnings exceeded expectations, leading to a surge in stock prices.": "positive",
    "The market is experiencing a downturn due to economic uncertainty and geopolitical tensions.": "negative",
    "The investment portfolio is diversified to mitigate risks and maximize returns.": "neutral",
    "The new product launch has been successful, resulting in increased revenue and market share.": "positive",
    "The financial report indicates a decline in profits, raising concerns among investors.": "negative"
}

guard = InputGuard()
llm_scanner = LLM(
            tag="default",
            instruction="""You are a finance expert. Determine if the given text is neutral or positive (valid) or negative (invalid) in sentiment, focusing on finance context.""",
            num_retry=2
            )
guard.add_scanner(llm_scanner)

corrects = []
for text, label in finance_examples.items():
    result = guard.scan(text)
    print(f"Expected Label: {label}, got {result.is_valid}")
    corrects.append((result.is_valid and label in ["neutral", "positive"]) or (not result.is_valid and label == "negative"))

acc = sum(corrects) / len(corrects)
print(f"Accuracy: {acc * 100:.2f}%")
```

```python
# topic classification for a random topic (e.g., space exploration)
random_topic_examples = {
    "NASA's Artemis program aims to return humans to the Moon.": True,
    "The James Webb Space Telescope has captured stunning images of distant galaxies.": True,
    "The stock market closed higher today after a volatile session.": False,
    "Baking a perfect sourdough bread requires a good starter and patience.": False,
    "SpaceX's Starship is designed for interplanetary travel to Mars.": True
}

guard = InputGuard()
topic_scanner = LLM(
    tag="default",
    instruction="""You are an expert in identifying topics. Determine if the given text is about space exploration (valid) or not (invalid).""",
    num_retry=2
)
guard.add_scanner(topic_scanner)

corrects = []
for text, is_about_topic in random_topic_examples.items():
    result = guard.scan(text)
    print(f"Expected Label: {is_about_topic}, got {result.is_valid}")
    corrects.append(result.is_valid == is_about_topic)

acc = sum(corrects) / len(corrects)
print(f"Accuracy: {acc * 100:.2f}%")
```
