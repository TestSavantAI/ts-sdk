```python
from testsavant.guard import InputGuard, OutputGuard
from testsavant.guard.output_scanners import MaliciousURLs
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
def dummy_llm(input):
    return input
long_article = "Once upon a time"
print(len(long_article))

ts_api = OutputGuard(
    API_KEY=os.environ.get("TEST_SAVANT_API_KEY"),
    PROJECT_ID=os.environ.get("TEST_SAVANT_PROJECT_ID"),
    remote_addr=os.environ.get("TEST_SAVANT_REMOTE_ADDR")
)
scanner = MaliciousURLs(tag="base", threshold=0.4)
ts_api.add_scanner(scanner)
```

```python
prompt = "como estas? quiero hablar de algo que me molesta mucho, retajconsultancy.com la gente que no respeta a los demás. http://www.juventudelirica.com.br/index.html"
prompts = [prompt, prompt + " " + long_article, long_article + " " + prompt]

for prompt in prompts:
    result = ts_api.scan(None, prompt)
    if result.is_valid:
        print("Prompt is valid")
    else:
        print("Prompt is invalid")
```
