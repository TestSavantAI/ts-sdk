```python
from testsavant.guard import InputGuard, Scanner
from testsavant.guard.input_scanners import Sentiment
import os
import dotenv
dotenv.load_dotenv()
```

```python
def dummy_llm(input):
    return input

reviews = [
    "I love this product! It has exceeded my expectations and I would highly recommend it to anyone.",
    "This is the worst experience I've ever had. The product was defective and the customer service was unhelpful. I will never buy from this company again.",
    "The product is okay, but the shipping took forever and the packaging was damaged. I expected better.",
    "I am extremely satisfied with my purchase. The quality is top-notch and the customer service was excellent."
]

ts_api = InputGuard()
```

```python
# detects of the sentiment of a text is negative
sentiment = Sentiment(tag="default", threshold=0.5) # a text with score more than 0.5 is considered negative sentiment
ts_api.add_scanner(sentiment)
```

```python
for review in reviews:
    result = ts_api.scan(review)
    print(f"Review: {review}")
    print(f"Is valid: {result.is_valid}")
    print(f"Sanitized prompt: {result.sanitized_prompt}")
    print("-" * 50)
```
